```python
%sql
-- Bronze layer: raw data from parquet files
-- match column stored as JSON STRING (parsed in silver layer)
CREATE TABLE IF NOT EXISTS workspace.futbol.bronze_matchs (
  scrape_date STRING,
  date STRING,
  name STRING,
  league STRING,
  link STRING,
  match STRING  -- JSON string from parquet
) USING DELTA;
```

```python
%sql
SELECT
    COUNT(*) as total_matchs,
    COUNT(DISTINCT league) AS total_leagues,
    MIN(date) AS min_date,
    MAX(date) AS last_date
FROM
    workspace.futbol.bronze_matchs
;
```

```python
%sql
COPY INTO workspace.futbol.bronze_matchs
FROM '/Volumes/workspace/futbol/futbol_landing'
FILEFORMAT = PARQUET
PATTERN = 'raw/year=*/month=*/day=*/*.parquet'
COPY_OPTIONS ('mergeSchema' = 'true');
```

```python
%sql
SELECT
    COUNT(*) as total_matchs,
    COUNT(DISTINCT league) AS total_leagues,
    MIN(date) AS min_date,
    MAX(date) AS last_date
FROM
    workspace.futbol.bronze_matchs
;
```
